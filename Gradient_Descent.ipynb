```python
def gradient_descent(x_true, y_true, epochs, loss_threshold = None):
    X = x_true.values
    y = y_true.values

    n_samples, n_features = X.shape

    w = np.ones(n_features)
    bias = 0
    rate = 0.5

    for i in range(epochs):
        weighted_sum = np.dot(X, w) + bias

        y_predicted = 1 / (1 + np.exp(-weighted_sum))

        loss = -(1/n_samples) * np.sum(
            y*np.log(y_predicted) + (1-y)*np.log(1-y_predicted)
        )

        dw = (1/n_samples) * np.dot(X.T, (y_predicted - y))
        db = np.mean(y_predicted - y)

        w -= rate * dw
        bias -= rate * db

        if loss_threshold is not None and loss <= loss_threshold:
          break

    return w, bias
```
